```python
import os
import re

# Example path (inside one checkpoint folder)
path = "/scratch/s223669184/project_data/Grant25/TimeSeriesECM/checkpoints/long_term_forecast_ECL_96_96_TimeXer_custom_ftM_sl96_ll48_pl96_dm512_nh8_el4_dl1_df512_expand2_dc4_fc3_ebtimeF_dtTrue_Exp_0/"

# Move one level up (the folder containing multiple checkpoints)
checkpoint_dir = os.path.dirname(path)

# Regex pattern to capture ECM type, coeff, and metric (MAE or MSE)
pattern = re.compile(r"checkpoint-modelerr-(\w+)-found-best-coeff-([\d.]+)-(MAE|MSE)")

best_coeffs = {}

# Loop over all files in the folder
for fname in os.listdir(checkpoint_dir):
    match = pattern.match(fname)
    if match:
        ecm_type = match.group(1)
        coeff_value = float(match.group(2))
        metric = match.group(3)

        if ecm_type not in best_coeffs:
            best_coeffs[ecm_type] = {}
        best_coeffs[ecm_type][metric] = coeff_value

# Print results
for ecm, metrics in best_coeffs.items():
    mae = metrics.get("MAE", "N/A")
    mse = metrics.get("MSE", "N/A")
    print(f"ECM: {ecm}, Best MAE Coeff: {mae}, Best MSE Coeff: {mse}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- Data ---
# Raw MSE for different datasets
etth1_mse = [0.4362, 0.4348, 0.4329, 0.4326, 0.4327, 0.4309, 0.4304, 0.4296]
weather_mse = [0.2414, 0.2396, 0.2395, 0.2469, 0.2408, 0.2388, 0.2393, 0.2452]
traffic_mse = [0.5485, 0.5534, 0.5485, 0.5477, 0.5483, 0.5482, 0.5450, 0.5432]

# Normalize MSE for first plot
etth1_norm = (np.array(etth1_mse) - min(etth1_mse)) / (max(etth1_mse) - min(etth1_mse))
weather_norm = (np.array(weather_mse) - min(weather_mse)) / (max(weather_mse) - min(weather_mse))
traffic_norm = (np.array(traffic_mse) - min(traffic_mse)) / (max(traffic_mse) - min(traffic_mse))

# Labels for x-axis
labels = ['No STD', '0.2-0.8', '0.3-0.7', '0.4-0.6',
          '0.5-0.5', '0.6-0.4', '0.7-0.3', '0.8-0.2']

# Weather improvements for different ST ratios (before -> after)
# Order per ST: TimeMixer, TimesNet, TimeXer
weather_before_after = {
    "0.5-0.5": [(0.2449, 0.2414), (0.2719, 0.2701), (0.2456, 0.2442)],
    "0.6-0.4": [(0.2449, 0.2388), (0.2719, 0.2675), (0.2456, 0.2417)],
    "0.7-0.3": [(0.2449, 0.2393), (0.2719, 0.2673), (0.2456, 0.2404)],
    "0.8-0.2": [(0.2449, 0.2452), (0.2719, 0.2670), (0.2456, 0.2421)]
}

# Compute % improvements
improvements = {}
for st, pairs in weather_before_after.items():
    improvements[st] = [(b - a) / b * 100 for b, a in pairs]

model_names = ["TimeMixer", "TimesNet", "TimeXer"]
x = np.arange(len(labels))  # x-axis for all 8 ST configurations
x2 = np.arange(len(model_names))
width = 0.18  # width of bars

# --- Plot ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# First subplot: Normalized MSE
axes[0].plot(x[1:], etth1_norm[1:len(x)], marker='o', label="ETTh1")
axes[0].plot(x[1:], weather_norm[1:len(x)], marker='s', label="Weather")
axes[0].plot(x[1:], traffic_norm[1:len(x)], marker='^', label="Traffic")
axes[0].set_xticks(x[1:])
axes[0].set_xticklabels(labels[1:len(x)], rotation=45)
# axes[0].set_xlabel("ST Configuration", fontsize=18)
axes[0].set_ylabel("Normalized MSE", fontsize=18)
axes[0].legend(fontsize=16, loc='upper center', bbox_to_anchor=(0.7, 1.0))  # slightly left
axes[0].grid(True)
axes[0].tick_params(axis='x', labelsize=18)  # x-axis tick labels
axes[0].tick_params(axis='y', labelsize=18)  # y-axis tick labels
axes[0].text(-0.15, 1.05, "(a)", transform=axes[0].transAxes, fontsize=20, fontweight='bold')
# Second subplot: % improvement grouped by ST
colors = ["skyblue", "orange", "green", "purple"]
hatches = ['/','-','\\', '|']  # different patterns for each bar

opacity = 0.7

for i, (st, vals) in enumerate(improvements.items()):
    axes[1].bar(x2 + i*width - width, vals, width=width, color=colors[i], edgecolor='black', label=f"{st}", alpha=opacity, hatch=hatches[i])

axes[1].set_xticks(x2 + width*0.5)  # center the ticks between grouped bars
axes[1].set_xticklabels(model_names, fontsize=18)
# axes[1].set_xlabel("Model Backbone", fontsize=14)
axes[1].set_ylabel("MSE Improvement (%)", fontsize=18)
# axes[1].set_title("Weather: % MSE Improvement by ST Ratio")
axes[1].grid(axis='y', linestyle="--", alpha=0.5)
axes[1].legend(fontsize=16)
axes[1].tick_params(axis='x', labelsize=18)  # x-axis tick labels
axes[1].tick_params(axis='y', labelsize=18)  # y-axis tick labels
axes[1].legend(fontsize=16, loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=4)  # For second subplot
axes[1].text(-0.15, 1.05, "(b)", transform=axes[1].transAxes, fontsize=20, fontweight='bold')

# Annotate bars
# for i, (st, vals) in enumerate(improvements.items()):
#     for j, v in enumerate(vals):
#         axes[1].text(j + i*width - width, v + 0.01, f"{v:.2f}%",  ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.savefig("my_plot.pdf", format="pdf")

plt.show()
```

### Coeficient Pattern Analysis

```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt

base_dir = "/home/s223669184/TimeSeriesECM"
models = ["TimeMixer", "TimesNet", "TimeXer"]

# Define a color map for datasets (add or adjust as needed)
dataset_colors = {
    "ETTh1": "red",
    "ETTh2": "blue",
    "ETTm1": "green",
    "ETTm2": "orange",
    "Traffic": "purple",
    "weather": "brown",
    # add other datasets here if needed
}

fig, axs = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

for i, model_name in enumerate(models):
    pattern = os.path.join(base_dir, f"None_None_*_{model_name}_best_metrics.csv")
    files = glob.glob(pattern)

    data = []
    for file in files:
        with open(file, 'r') as f:
            mse_None = 0.0
            for line in f:

                parts = line.strip().split(',')
                method, loss, coef, mse, *_ = parts
                if method == "NA":
                    mse_None = float(mse)

                if len(parts) < 5:
                    continue
                method, loss, coef, mse, *_ = parts
                if loss == "MSE" and method == "TF":
                    dataset_name = file.split('/')[-1].replace("None_None_", "").replace("_best_metrics.csv", "")
                    dataset_name = "_".join(dataset_name.split("_")[:-1])  # Remove last part (model name)
                    data.append({
                        "Dataset": dataset_name,
                        "Coef": float(coef),
                        "MSE": float(mse_None)
                    })

    df = pd.DataFrame(data)
    if df.empty:
        print(f"No data found for model {model_name}")
        continue
    print(df)
    ax = axs[i]
    for dataset in df['Dataset'].unique():
        subset = df[df['Dataset'] == dataset].sort_values(by='Coef')
        color = dataset_colors.get(dataset, None)  # Use predefined color or None for default
        ax.plot(subset['Coef'], subset['MSE'], 'o-', label=dataset, color=color)

    ax.set_title(model_name)
    ax.set_xlabel("Best_Coef")
    ax.set_xticks([0.1, 0.3, 0.5, 0.7, 1])  # <-- set x-axis ticks here

    if i == 0:
        ax.set_ylabel("MSE")
    ax.grid(True)
    ax.legend(title="Dataset", fontsize=8)

plt.suptitle("Best Coefficient vs MSE for TF Method (per Model and Dataset)")
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
import os
import re
import glob
import pandas as pd

# Path pattern for all result files
path_pattern = "/home/s223669184/TimeSeriesECM/*_best_metrics.csv"
files = glob.glob(path_pattern)

# Pattern to extract seasonal, trend, dataset, model
pattern = re.compile(r'(?P<seasonal>\d\.\d+)_(?P<trend>\d\.\d+)_(?P<dataset>[^_]+)_(?P<model>[^_]+)_best_metrics\.csv')

# Store best results
best_results = {}

for filepath in files:
    filename = os.path.basename(filepath)
    match = pattern.match(filename)
    if not match:
        print(f"Skipping file due to filename mismatch: {filename}")
        continue

    try:
        # Extract info
        seasonal = float(match.group("seasonal"))
        trend = float(match.group("trend"))
        dataset = match.group("dataset")
        model = match.group("model")

        df = pd.read_csv(filepath)

        df = df[df["Best_Coef"] != 0.0]


        # Loop through rows to find best MSE and MAE
        for _, row in df.iterrows():
            key_mse = f"{dataset}_{model}_MSE"
            key_mae = f"{dataset}_{model}_MAE"

            mse = float(row.get("Avg_MSE", float("inf")))
            mae = float(row.get("Avg_MAE", float("inf")))

            # Update best MSE
            if key_mse not in best_results or mse < best_results[key_mse]['value']:
                best_results[key_mse] = {
                    "value": mse,
                    "seasonal": seasonal,
                    "trend": trend,
                    "coef": row.get("Best_Coef", None),
                    "method": row.get("Method", None)
                }

            # Update best MAE
            if key_mae not in best_results or mae < best_results[key_mae]['value']:
                best_results[key_mae] = {
                    "value": mae,
                    "seasonal": seasonal,
                    "trend": trend,
                    "coef": row.get("Best_Coef", None),
                    "method": row.get("Method", None)
                }

    except Exception as e:
        print(f"Error reading {filename}: {e}")

# Display best results
print("\nBest results summary (lowest MSE/MAE per dataset/model):")
summary_data = []
for key, info in best_results.items():
    dataset, model, metric = key.split("_")
    summary_data.append({
        "Dataset": dataset,
        "Model": model,
        "Metric": metric,
        "Best_Value": info["value"],
        "Seasonal": info["seasonal"],
        "Trend": info["trend"],
        "Best_Coef": info["coef"],
        "Method": info["method"]
    })

summary_df = pd.DataFrame(summary_data)
print(summary_df)

# Optionally save to CSV
summary_df.to_csv("best_metric_summary.csv", index=False)
```
